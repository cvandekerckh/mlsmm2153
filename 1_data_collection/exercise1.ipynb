## Step 3: XML files

An XML file is a **tree of tags**: each piece of information is enclosed between an opening tag `<name>` and a closing tag `</name>`, and tags can contain other tags. Open `clinical_trials.xml` in a text editor to see what it looks like:

```xml
<clinical_study>
  <brief_title>Isolation and Culture of Immune Cells ...</brief_title>
  ...
</clinical_study>
```

With `ElementTree`, we read the file and get the **root** of the tree (here `<clinical_study>`):

```python
root = ET.parse(path).getroot()
root.find("tag_name")   # returns the first child tag with this name
root.find("tag_name").text   # returns the text it contains
```

## Step 4: HTML files

An HTML file is a web page. Like XML, it is a tree of tags (`<title>`, `<div>`, `<a>`, `<table>`, ...), but real web pages are often messy, so we use a more tolerant tool: **BeautifulSoup**.

```python
soup = BeautifulSoup(html_text, "html.parser")
soup.title.text           # the text of the <title> tag
soup.find_all("a")        # the list of all <a> tags (the links)
```
